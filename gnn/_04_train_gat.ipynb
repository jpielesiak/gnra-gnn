# GAT RNA Motif Classifier (v9)

Graph Attention Network (GATv2) for GNRA tetraloop detection.

This notebook follows the **same pipeline** as `10-remove-redundancy-and-train-models.ipynb`:
- Input: `geometric_features.csv`
- Raw angle columns are dropped (sin/cos variants are kept)
- Splits are built from the same 6 clustering JSON files
  (approximate/exact × hierarchical/affinity-propagation/facility-location)
- One model is trained and saved per split

What differs from classical ML:
- Features are converted into graphs (distances → edges, sin/cos angles → node features)
- The classifier is a 3-layer GATv2 with global mean pooling

**Additional data hygiene (new):**
- **Temporal split**: training uses only structures released *before* `CUTOFF_DATE`;
  the test set uses structures released *on or after* it — preventing data leakage.
- **Cluster de-duplication**: for each redundancy cluster only `n_splits` representatives
  are kept (one per CV fold), so no cluster contributes members to both train and test.


**v6 change:** fixes a train/test leakage bug in the cluster de-duplication step. In v5, de-duplication was run *separately* on the pre-cutoff and post-cutoff pools, so a redundancy cluster could still contribute near-duplicate structures to both train and test. v6 runs de-duplication jointly across both pools: any cluster straddling the cutoff has its train-side members dropped entirely, guaranteeing the test pool is never contaminated by something the model could have memorized during training.

**v7 change:** v6's de-duplication was a silent no-op — it looked for a file called `clusters.json`, which doesn't exist (the real per-split cluster files are `{mode}-{method}.json`, already loaded into the `clustering` dict in the Cluster Loading cell). It also computed de-duplication once globally, before the main loop, even though the loop iterates over 6 genuinely different clusterings — so all six "splits" were silently training on identical data. v7 fixes both: de-duplication now runs **once per split, inside the main loop**, using that split's own real cluster list, and still keeps the cross-pool leak fix (any cluster straddling the cutoff has its train-side members dropped).

**v8 change:** v7's de-duplication was confirmed working (different, real train/test sizes per split) — yet every split still scored 1.00 across the board. That ruled out redundant rows as the (sole) explanation. The actual cause: the Main Training Loop used `test_loader` — the held-out post-cutoff set — every epoch to pick the "best" checkpoint (early stopping on test MCC), then reported that same checkpoint's score on that same test set. That guarantees inflated numbers no matter how clean the row-level split is, because the reported "test" score is really the score of whichever epoch happened to fit the test set best. v8 fixes this: a validation set is now carved out of the *train* pool for early stopping, and the test set is touched exactly once, after training is fully finished, purely to report the final metric.

## Cluster Loading

Identical to `10-remove-redundancy-and-train-models.ipynb`.

```python
class Cluster:
    def __init__(self, representative, members):
        self.representative = representative
        self.members = members

    def __repr__(self):
        return (
            f"Cluster(representative='{self.representative}', members={self.members})"
        )


def clean_name(name: str) -> str:
    return name.split("/")[-1].split(".")[0]


def load_clusters(array) -> list:
    result = []
    for obj in array:
        representative = clean_name(obj["representative"])
        members = list(map(clean_name, obj["members"]))
        result.append(Cluster(representative, members))
    return result


def load_all_clusters():
    for mode in ["approximate", "exact"]:
        for method in ["hierarchical", "affinity-propagation", "facility-location"]:
            path = f"{mode}-{method}.json"
            with open(path) as f:
                data = json.load(f)
                yield (mode, method, load_clusters(data["clustering"]["clusters"]))


clustering = {}
for mode, method, clusters in load_all_clusters():
    clustering[(mode, method)] = clusters

# Sanity-check cluster counts (same assertions as notebook 10)
assert len(clustering[("approximate", "hierarchical")]) == 88
assert len(clustering[("approximate", "affinity-propagation")]) == 34
assert len(clustering[("approximate", "facility-location")]) == 33
assert len(clustering[("exact", "hierarchical")]) == 82
assert len(clustering[("exact", "affinity-propagation")]) == 34
assert len(clustering[("exact", "facility-location")]) == 33

print(f"Loaded {len(clustering)} splits.")
```

## Date-Based Temporal Split

All structures are split by PDB release date.  Only structures released
**before** `CUTOFF_DATE` enter the training pool; those released on or
after it form the held-out test pool.  This prevents temporal data leakage.


```python
def filter_by_date(df, dates_csv, cutoff_date):
    """
    Split *df* into (pre_df, post_df) by PDB release date.

    The DataFrame index or 'source_file' column must encode the PDB ID as
    the 2nd underscore-delimited token, e.g. 'HL_1G1X_001'.
    4-token names like 'single_strands_2il9_0003' use the 3rd token.

    Returns
    -------
    pre_df   : rows with release_date <  cutoff_date  (training)
    post_df  : rows with release_date >= cutoff_date  (test/holdout)
    """
    cutoff = pd.to_datetime(cutoff_date, utc=True)

    release_dates = {}
    with open(dates_csv, 'r') as f:
        for row in csv.DictReader(f):
            release_dates[row['pdbid']] = pd.to_datetime(row['release_date'], utc=True)

    def _pdbid(src):
        parts = src.split('_')
        return parts[2] if len(parts) == 4 else (parts[1] if len(parts) >= 2 else src)

    pre_rows, post_rows, lost = [], [], []
    use_index = 'source_file' not in df.columns

    for idx, row in df.iterrows():
        src   = str(idx) if use_index else str(row['source_file'])
        pdbid = _pdbid(src)
        rd    = (release_dates.get(pdbid)
                 or release_dates.get(pdbid.upper())
                 or release_dates.get(pdbid.lower()))
        if rd is None:
            lost.append(pdbid)
            continue
        (pre_rows if rd < cutoff else post_rows).append(row)

    pre_df  = pd.DataFrame(pre_rows,  columns=df.columns)
    post_df = pd.DataFrame(post_rows, columns=df.columns)

    print(f'Total entries        : {len(df)}')
    print(f'Before cutoff (train): {len(pre_df)}')
    print(f'After  cutoff (test) : {len(post_df)}')
    if lost:
        shown = lost[:10]
        ellipsis = '...' if len(lost) > 10 else ''
        print(f'PDB IDs not in dates CSV ({len(lost)}): {shown}{ellipsis}')
    return pre_df, post_df


# Apply temporal filter
df_pre_cutoff, df_post_cutoff = filter_by_date(df_filtered, DATES_CSV, CUTOFF_DATE)
print()
print(f'Positives in train pool : {df_pre_cutoff["gnra"].sum()}')
print(f'Positives in test  pool : {df_post_cutoff["gnra"].sum()}')
```

## Cluster De-duplication

RNA structures are grouped into redundancy clusters.  If all members of a
cluster were kept, the same structural motif could appear in both train and
test, inflating apparent performance.

**v7 fix:** v6 pointed at a nonexistent `clusters.json` file, so de-duplication silently never ran (v5 had the same problem). The real cluster data for each split already lives in the `clustering` dict loaded above (`clustering[(mode, method)]`). De-duplication must also be split-specific — each of the 6 clusterings groups structures differently — so it can't be done once globally before the loop; it now runs **inside the Main Training Loop**, once per split:
- Clusters entirely inside one pool: keep only 1 representative (redundant members dropped).
- Clusters straddling the cutoff (members in both pools): the post-cutoff member(s) are kept as the real holdout; **all** train-side members of that cluster are dropped, so no near-duplicate of a test example survives in training.


```python
def dedup_clusters_for_split(df_pre, df_post, clusters, n_keep=1, seed=SEED):
    """
    Cluster-aware, train/test-safe de-duplication for a SINGLE clustering split.

    v6 bug this replaces: the dedup call pointed at a hardcoded 'clusters.json'
    file that does not exist anywhere in this pipeline — every real cluster
    file is named '{mode}-{method}.json' and is already loaded into the
    `clustering` dict in the Cluster Loading cell. Because the file was never
    found, de-duplication silently no-op'd (see the "clusters.json not found —
    skipping" message you saw), for both v5 and v6. No redundancy was ever
    actually removed.

    v6 also had a second, separate problem: de-duplication was computed once,
    globally, before the main loop — but the main loop iterates over 6
    genuinely different clusterings (`clustering[(mode, method)]`). Reusing one
    global dedup for all six meant every "split" trained on identical data,
    which is why all six reported identical sample counts. De-duplication has
    to be redone per split, using that split's own cluster membership.

    Parameters
    ----------
    df_pre, df_post : the date-filtered (NOT yet deduplicated) train/test pools
    clusters         : list[Cluster] for this one (mode, method) split
    n_keep           : max representatives kept per cluster, within a pool
                       (default 1 — a single split doesn't need multiple CV-fold
                       representatives, unlike the old cross-split reuse logic)

    Cross-pool safety (unchanged from the earlier fix): if a cluster has
    members in BOTH df_pre and df_post, the post (test) side is treated as the
    real holdout and left alone; ALL of that cluster's train-side members are
    dropped instead, so no near-duplicate of a test example survives in
    training.
    """
    use_index = 'source_file' not in df_pre.columns
    pre_ids  = set(df_pre.index.astype(str))  if use_index else set(df_pre['source_file'].astype(str))
    post_ids = set(df_post.index.astype(str)) if use_index else set(df_post['source_file'].astype(str))

    rng = np.random.default_rng(seed)
    pre_drop, post_drop = set(), set()
    n_cross = n_pre_only = n_post_only = 0

    for cl in clusters:
        members = list(cl.members)
        in_pre  = [m for m in members if m in pre_ids]
        in_post = [m for m in members if m in post_ids]

        if in_pre and in_post:
            n_cross += 1
            pre_drop.update(in_pre)          # drop entire cluster from train
            rng.shuffle(in_post)
            keep = set(in_post[:n_keep])
            post_drop.update(m for m in in_post if m not in keep)
        elif in_pre:
            n_pre_only += 1
            rng.shuffle(in_pre)
            keep = set(in_pre[:n_keep])
            pre_drop.update(m for m in in_pre if m not in keep)
        elif in_post:
            n_post_only += 1
            rng.shuffle(in_post)
            keep = set(in_post[:n_keep])
            post_drop.update(m for m in in_post if m not in keep)

    if use_index:
        deduped_pre  = df_pre[~df_pre.index.astype(str).isin(pre_drop)]
        deduped_post = df_post[~df_post.index.astype(str).isin(post_drop)]
    else:
        deduped_pre  = df_pre[~df_pre['source_file'].astype(str).isin(pre_drop)]
        deduped_post = df_post[~df_post['source_file'].astype(str).isin(post_drop)]

    info = {
        'clusters_total':      len(clusters),
        'cross_pool_clusters': n_cross,
        'pre_only_clusters':   n_pre_only,
        'post_only_clusters':  n_post_only,
        'rows_removed_pre':    len(df_pre)  - len(deduped_pre),
        'rows_removed_post':   len(df_post) - len(deduped_post),
    }
    return deduped_pre, deduped_post, info


# NOTE: this function is intentionally NOT called here. De-duplication depends
# on which of the 6 clusterings is being used, so it now runs once per
# iteration inside the Main Training Loop below, using that split's own
# `clustering[(mode, method)]` list — see the loop for the actual call and
# the per-split de-dup counts it prints.
print("dedup_clusters_for_split() defined. Runs per-split inside the main "
      "training loop (each of the 6 clusterings gets its own de-duplication).")
```

## Graph Encoding

Converts a flat feature row into a `torch_geometric.data.Data` object.

Mapping rules (derived from column-name digit parsing, 0-indexed):
- 2 digit indices → **edge** feature (distance `d{i}{j}`)
- 3 digit indices → **node** feature for the *middle* node (`sin_a{i}{j}{k}`, `cos_a{i}{j}{k}`)
- 4 digit indices → **node** feature for *both* middle nodes (`sin_t{i}{j}{k}{l}`, `cos_t{i}{j}{k}{l}`)

There is no nucleotide sequence in `geometric_features.csv`, so node features are
purely geometric (angle sin/cos). Terminal nodes (0 and 7) receive zero-padded
features because they never appear as angle/torsion midpoints.

Edges are made **undirected** by mirroring `(i→j)` and `(j→i)`.

```python
def _digits(col: str) -> list[int]:
    """Extract all digit characters from a column name as a list of ints."""
    return [int(c) for c in col if c.isdigit()]


def row_to_graph(row: pd.Series, feat_cols: list[str]) -> Data:
    """
    Convert one row of the (scaled) feature DataFrame into a PyG Data object.

    Parameters
    ----------
    row       : a row from df_filtered (must already be scaled)
    feat_cols : list of feature column names (excludes 'source_file' and 'gnra')
    """
    edge_dict    = {}                            # (i, j) -> [distance, ...]
    node_angle   = {i: [] for i in range(NUM_NODES)}
    node_torsion = {i: [] for i in range(NUM_NODES)}

    for col in feat_cols:
        idxs = _digits(col)
        val  = float(row[col]) if not pd.isna(row[col]) else 0.0

        if len(idxs) == 2:
            i, j = idxs[0], idxs[1]
            # Guard against indices that exceed the window (shouldn't happen with 0-7)
            if i < NUM_NODES and j < NUM_NODES:
                edge_dict.setdefault((i, j), []).append(val)

        elif len(idxs) == 3:
            middle = idxs[1]          # the middle atom carries the angle feature
            if all(x < NUM_NODES for x in idxs):
                node_angle[middle].append(val)

        elif len(idxs) == 4:
            mid1, mid2 = idxs[1], idxs[2]
            if all(x < NUM_NODES for x in idxs):
                node_torsion[mid1].append(val)
                node_torsion[mid2].append(val)

    # ── Node feature matrix ───────────────────────────────────────────────────
    max_angle   = max((len(v) for v in node_angle.values()),   default=0)
    max_torsion = max((len(v) for v in node_torsion.values()), default=0)

    node_feats = []
    for i in range(NUM_NODES):
        a = node_angle[i]   + [0.0] * (max_angle   - len(node_angle[i]))
        t = node_torsion[i] + [0.0] * (max_torsion - len(node_torsion[i]))
        node_feats.append(a + t)

    x = torch.tensor(node_feats, dtype=torch.float32)   # shape [NUM_NODES, node_dim]

    # ── Edge index + attributes ───────────────────────────────────────────────
    edge_index_list: list = []
    edge_attr_list:  list = []

    for (i, j), weights in edge_dict.items():
        is_consecutive = 1.0 if abs(i - j) == 1 else 0.0
        # Keep only the first accumulated value (single distance per pair)
        # plus the is_consecutive flag → EDGE_DIM = 2
        edge_attr_list.append([weights[0], is_consecutive])
        edge_index_list.append([i, j])

    if edge_index_list:
        edge_attr  = torch.tensor(edge_attr_list,  dtype=torch.float32)
        edge_index = torch.tensor(edge_index_list, dtype=torch.int64).t().contiguous()
        # Make undirected
        edge_index = torch.cat([edge_index, edge_index.flip(0)], dim=1)
        edge_attr  = torch.cat([edge_attr,  edge_attr],          dim=0)
    else:
        edge_index = torch.zeros((2, 0), dtype=torch.int64)
        edge_attr  = torch.zeros((0, EDGE_DIM), dtype=torch.float32)

    y = torch.tensor([int(row["gnra"])], dtype=torch.int64)
    return Data(x=x, edge_index=edge_index, edge_attr=edge_attr, y=y)
```

## GATv2 Model Definition

```python
class GATModel(torch.nn.Module):
    """
    3-layer GATv2 graph classifier.

    Architecture
    ------------
    conv1 : GATv2Conv(NODE_DIM  → hidden1,  heads=heads,  concat=True)
    conv2 : GATv2Conv(hidden1*h → hidden2,  heads=heads,  concat=True)
    conv3 : GATv2Conv(hidden2*h → hidden3,  heads=1,      concat=False)
    pool  : global mean pool  →  shape [batch, hidden3]
    lin   : Linear(hidden3, 2)

    hp keys
    -------
    hidden1, hidden2, hidden3 : layer output channels
    heads                     : attention heads (shared across conv1/conv2)
    dropout                   : dropout before the linear classifier
    """

    def __init__(self, hp: dict, node_dim: int, edge_dim: int):
        super().__init__()
        h1, h2, h3 = hp["hidden1"], hp["hidden2"], hp["hidden3"]
        heads = hp["heads"]

        self.conv1 = GATv2Conv(node_dim,    h1, edge_dim=edge_dim, heads=heads,  concat=True)
        self.conv2 = GATv2Conv(h1 * heads,  h2, edge_dim=edge_dim, heads=heads,  concat=True)
        self.conv3 = GATv2Conv(h2 * heads,  h3, edge_dim=edge_dim, heads=1,      concat=False)
        self.lin   = torch.nn.Linear(h3, 2)
        self.dropout_p = hp["dropout"]

    def forward(self, x, edge_index, edge_attr, batch):
        x = self.conv1(x, edge_index, edge_attr=edge_attr).relu()
        x = self.conv2(x, edge_index, edge_attr=edge_attr).relu()
        x = self.conv3(x, edge_index, edge_attr=edge_attr).relu()
        x = global_mean_pool(x, batch)
        x = F.dropout(x, p=self.dropout_p, training=self.training)
        return self.lin(x)
```

## Training & Evaluation Helpers

```python
def train_epoch(model, loader, optimizer, criterion) -> None:
    model.train()
    for data in loader:
        optimizer.zero_grad()
        out  = model(data.x, data.edge_index, data.edge_attr, data.batch)
        loss = criterion(out, data.y)
        loss.backward()
        optimizer.step()


def evaluate(model, loader) -> tuple[float, np.ndarray, np.ndarray]:
    """Return (accuracy, predictions, labels) for the given DataLoader."""
    model.eval()
    all_preds, all_labels = [], []
    correct = 0

    with torch.no_grad():
        for data in loader:
            out  = model(data.x, data.edge_index, data.edge_attr, data.batch)
            pred = out.argmax(dim=1)
            correct += int((pred == data.y).sum())
            all_preds.extend(pred.cpu().numpy().tolist())
            all_labels.extend(data.y.cpu().numpy().tolist())

    accuracy = correct / len(loader.dataset)
    return accuracy, np.array(all_preds), np.array(all_labels)


def compute_metrics(labels: np.ndarray, preds: np.ndarray) -> dict:
    """Compute accuracy, precision, recall, F1 (macro), and MCC."""
    return {
        "accuracy":  accuracy_score(labels, preds),
        "precision": precision_score(labels, preds, zero_division=0),
        "recall":    recall_score(labels, preds, zero_division=0),
        "f1":        f1_score(labels, preds, zero_division=0),
        "mcc":       matthews_corrcoef(labels, preds),
    }


def save_training_chart(
    history: dict,
    split_label: str,
    out_dir: str = "charts",
) -> str:
    """
    Save a 5-panel PNG with per-epoch training curves.

    Parameters
    ----------
    history    : dict with keys 'accuracy', 'precision', 'recall', 'f1', 'mcc',
                 each mapping to a list of per-epoch float values.
    split_label: human-readable label used in the title and filename.
    out_dir    : directory where the PNG is written.

    Returns
    -------
    Path to the saved PNG.
    """
    metrics = ["accuracy", "precision", "recall", "f1", "mcc"]
    titles  = ["Accuracy", "Precision", "Recall", "F1 Score", "MCC"]
    colors  = ["#2196F3", "#4CAF50", "#FF9800", "#9C27B0", "#F44336"]

    epochs = list(range(1, len(history["accuracy"]) + 1))

    fig, axes = plt.subplots(1, 5, figsize=(20, 4), constrained_layout=True)
    fig.suptitle(f"Training Curves — {split_label}", fontsize=13, fontweight="bold")

    for ax, metric, title, color in zip(axes, metrics, titles, colors):
        ax.plot(epochs, history[metric], color=color, linewidth=1.8)
        best_val = max(history[metric])
        best_ep  = history[metric].index(best_val) + 1
        ax.axvline(best_ep, color=color, linestyle="--", alpha=0.4,
                   label=f"Best: {best_val:.3f} (ep {best_ep})")
        ax.set_title(title, fontweight="bold")
        ax.set_xlabel("Epoch")
        ax.set_ylim(-0.05, 1.05) if metric != "mcc" else ax.set_ylim(-1.05, 1.05)
        ax.legend(fontsize=8)
        ax.grid(True, alpha=0.3)

    safe_label = split_label.replace(" / ", "-").replace(" ", "_")
    out_path   = os.path.join(out_dir, f"training_{safe_label}.png")
    fig.savefig(out_path, dpi=120)
    plt.close(fig)
    return out_path
```

## Optuna Hyperparameter Optimisation

Defines the CV objective (`make_objective`) and a `run_optuna_for_split()`
helper. Neither runs a study here.

**v9 fix:** this used to run **once**, on the raw pre-cutoff pool (before
any de-duplication), and the resulting HP dict was reused for all six
cluster-based splits. That's a mismatch on two levels:
- Each split's de-duplication removes a *different* set of near-duplicate
  rows, so tuning on the raw pool means Optuna's CV folds could contain
  near-identical motifs split across train/val — letting it learn
  hyperparameters that exploit that duplication rather than generalise.
- Whatever HP came out of that single study was then reused for every split,
  even though each split's de-duplicated training distribution is different.

Now the actual `optuna.create_study(...).optimize(...)` call happens **once
per split**, inside the Main Training Loop below, using that split's own
already-deduplicated pre-cutoff pool. De-dup → Optuna → final training all
run back-to-back for each split, so the hyperparameters are tuned on exactly
the distribution the final model is trained on.


```python
from sklearn.model_selection import StratifiedKFold

# ── Optuna objective (CV over a split's own de-duplicated train pool) ──────────
def make_objective(df_hpo, feat_cols, node_dim, edge_dim, n_splits=5):
    """
    Returns a closure that Optuna calls once per trial.
    For each trial: runs stratified k-fold CV *within the pool it's given*
    (the caller passes in that split's already-deduplicated pre-cutoff pool),
    trains a short run per fold, and returns mean(fold MCC) - 0.5*std(fold MCC).
    The post-cutoff test set is NEVER touched here — it stays a clean holdout
    for the final training loop.
    """
    labels = df_hpo["gnra"].astype(int).values
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=SEED)
    fold_indices = list(skf.split(df_hpo, labels))  # fixed across trials

    def objective(trial: optuna.Trial) -> float:
        hp = {
            "hidden1":      trial.suggest_categorical("hidden1",      [16, 32, 64]),
            "hidden2":      trial.suggest_categorical("hidden2",      [32, 64, 128]),
            "hidden3":      trial.suggest_categorical("hidden3",      [16, 32, 64]),
            "heads":        trial.suggest_categorical("heads",        [1, 2, 4]),
            "dropout":      trial.suggest_float("dropout",            0.1, 0.5, step=0.05),
            "lr":           trial.suggest_float("lr",                 1e-4, 1e-2, log=True),
            "weight_decay": trial.suggest_float("weight_decay",       0.0,  1e-3, step=1e-4),
            "batch_size":   HP_BATCH,
            "max_epochs":   OPTUNA_MAX_EPOCHS,
            "patience":     OPTUNA_PATIENCE,
        }

        fold_mccs = []

        for fold_i, (tr_idx, val_idx) in enumerate(fold_indices):
            fold_train_df = df_hpo.iloc[tr_idx].reset_index(drop=True)
            fold_val_df   = df_hpo.iloc[val_idx].reset_index(drop=True)

            # Fit scaler on this fold's train split only — no leakage across folds
            fold_scaler = StandardScaler()
            fold_train_df = fold_train_df.copy()
            fold_val_df   = fold_val_df.copy()
            fold_train_df[feat_cols] = fold_scaler.fit_transform(fold_train_df[feat_cols])
            fold_val_df[feat_cols]   = fold_scaler.transform(fold_val_df[feat_cols])

            fold_train_ds = [row_to_graph(row, feat_cols) for _, row in fold_train_df.iterrows()]
            fold_val_ds   = [row_to_graph(row, feat_cols) for _, row in fold_val_df.iterrows()]

            fold_train_loader = DataLoader(fold_train_ds, batch_size=HP_BATCH, shuffle=True)
            fold_val_loader   = DataLoader(fold_val_ds,   batch_size=HP_BATCH)

            n_pos = int(fold_train_df["gnra"].sum())
            n_neg = int((~fold_train_df["gnra"]).sum())
            pos_weight    = n_neg / n_pos if n_pos > 0 else 1.0
            class_weights = torch.tensor([1.0, pos_weight], dtype=torch.float)
            criterion     = torch.nn.CrossEntropyLoss(weight=class_weights)

            # Distinct seed per (trial, fold) so folds aren't identical runs
            set_all_seeds(SEED + trial.number * 100 + fold_i)
            model     = GATModel(hp, node_dim, edge_dim)
            optimizer = torch.optim.Adam(
                model.parameters(), lr=hp["lr"], weight_decay=hp["weight_decay"]
            )

            best_mcc   = -2.0
            no_improve = 0

            for epoch in range(1, OPTUNA_MAX_EPOCHS + 1):
                train_epoch(model, fold_train_loader, optimizer, criterion)
                _, val_preds, val_labels = evaluate(model, fold_val_loader)
                val_mcc = matthews_corrcoef(val_labels, val_preds)

                if val_mcc > best_mcc:
                    best_mcc   = val_mcc
                    no_improve = 0
                else:
                    no_improve += 1

                if no_improve >= OPTUNA_PATIENCE:
                    break

            fold_mccs.append(best_mcc)

            # Report running mean after each fold so the pruner has something
            # to look at; still no TrialPruned raised — same policy as before.
            trial.report(float(np.mean(fold_mccs)), fold_i)

        mean_mcc = float(np.mean(fold_mccs))
        std_mcc  = float(np.std(fold_mccs))
        # Penalize instability across folds, not just chase the mean
        return mean_mcc - 0.5 * std_mcc

    return objective


# ── Default HP (used as fallback if every trial is pruned) ─────────────────────
HP_DEFAULT = {
    "hidden1":      32,
    "hidden2":      64,
    "hidden3":      32,
    "heads":         2,
    "dropout":      0.3,
    "lr":           1e-3,
    "weight_decay": 0.0,
}

HP_BATCH     = 32
HPO_N_SPLITS = 5   # stratified folds within each split's de-duplicated pool


def run_optuna_for_split(df_hpo, feat_cols, node_dim, edge_dim, split_label,
                          n_trials=OPTUNA_N_TRIALS, n_splits=HPO_N_SPLITS, seed=SEED):
    """
    Runs ONE fresh Optuna study for ONE split, using that split's own
    de-duplicated pre-cutoff pool (`df_hpo`) for the CV objective.

    v9 fix this replaces: Optuna used to tune once, globally, on the RAW
    (not-yet-deduplicated) pre-cutoff pool, and the resulting HP dict was
    reused for every one of the 6 splits. Because de-duplication is
    cluster/split-specific, that raw pool could contain near-identical
    duplicate motifs (same cluster, both copies present) that don't exist in
    the deduplicated data any split actually trains its final model on.
    Tuning on that raw, duplicate-containing pool risked Optuna favouring
    hyperparameters that merely exploit the duplication in its own CV folds —
    a different distribution than the deduplicated data used downstream.

    Now: for each split, de-dup runs first, THEN this function tunes on that
    split's deduplicated pool, THEN the final model for that split trains
    with the HP this study found. All three steps run back-to-back per split
    on consistent data, instead of "tune once -> reuse everywhere".

    Returns a complete HP dict (best trial's params + fixed batch/epoch/
    patience settings), ready to hand to the final per-split training loop.
    """
    study = optuna.create_study(
        direction="maximize",
        sampler=optuna.samplers.TPESampler(seed=seed),
        pruner=optuna.pruners.PatientPruner(optuna.pruners.NopPruner(), patience=OPTUNA_PATIENCE),
    )
    study.optimize(
        make_objective(df_hpo, feat_cols, node_dim, edge_dim, n_splits=n_splits),
        n_trials=n_trials,
        show_progress_bar=True,
    )

    completed = [t for t in study.trials if t.state == optuna.trial.TrialState.COMPLETE]

    if completed:
        best_trial = max(completed, key=lambda t: t.value)
        print(f"  Optuna [{split_label}] best trial #{best_trial.number}  |  "
              f"CV score (mean MCC - 0.5*std) = {best_trial.value:.4f}")
        for k, v in best_trial.params.items():
            print(f"      {k:>15s} = {v}")
        best_params = best_trial.params
    else:
        print(f"  Optuna [{split_label}]: no trials completed (all pruned). "
              f"Falling back to default HP.")
        best_params = HP_DEFAULT

    return {
        **best_params,
        "batch_size": HP_BATCH,
        "max_epochs": FINAL_MAX_EPOCHS,
        "patience":   FINAL_PATIENCE,
    }


print("Optuna objective + run_optuna_for_split() defined. A fresh study now "
      "runs once per split, inside the Main Training Loop, on that split's "
      "own de-duplicated pre-cutoff pool — see the loop below for the call "
      "and the per-split best-trial printout.")
```

## Main Training Loop

Iterates over all 6 cluster-based splits. For **each** split, in order:

1. **De-duplicate** this split's pre/post cutoff pools using its own cluster
   list (v7 fix).
2. **Tune Optuna on THIS split's deduplicated pool** and get an HP dict back
   (v9 fix — previously one global HP dict, tuned once on the *raw,
   non-deduplicated* pool, was reused for every split; now each split gets
   its own fresh study on exactly the data it will train on).
3. Carve a validation set out of the deduplicated train pool (v8 fix) and
   train the final model with that split's own HP, early-stopping on
   validation MCC. The held-out **test set is used exactly once**, after
   training finishes, purely to report the final metrics — never during
   training or model selection, and never during Optuna tuning either.

Per-epoch **validation** metrics (accuracy, precision, recall, F1, MCC) are
recorded and saved as PNG charts in `charts/`. Each split's tuned HP is
stashed in `HP_by_split[(mode, method)]` so later cells (e.g. the
label-shuffle check) can reuse the right HP for the right split.


```python
set_all_seeds(SEED)

summary_rows  = []   # collects final metrics per split for a summary table
HP_by_split   = {}   # (mode, method) -> that split's own Optuna-tuned HP dict

for (mode, method), clusters in clustering.items():
    print(f"\n{'='*60}")
    print(f"  Split: {mode} / {method}")
    print(f"{'='*60}")

    # ── Build train/test split (temporal + de-duplicated FOR THIS SPLIT) ──────
    # Training data : structures released before CUTOFF_DATE.
    # Test data     : structures released on/after CUTOFF_DATE.
    # De-duplication uses THIS split's own cluster list (`clusters`), applied
    # to the raw date-filtered pools (df_pre_cutoff / df_post_cutoff — these
    # are untouched by any prior global dedup). Cross-pool leakage is handled
    # inside dedup_clusters_for_split: any cluster with members on both sides
    # of the cutoff has its train-side members dropped so the test pool for
    # THIS split is never contaminated by a near-duplicate the model trained on.
    df_pre_split, df_post_split, dedup_info = dedup_clusters_for_split(
        df_pre_cutoff, df_post_cutoff, clusters, n_keep=1
    )
    print(f"  De-dup: {dedup_info['clusters_total']} clusters "
          f"({dedup_info['cross_pool_clusters']} crossed train/test, "
          f"{dedup_info['pre_only_clusters']} train-only, "
          f"{dedup_info['post_only_clusters']} test-only)")
    print(f"  De-dup: removed {dedup_info['rows_removed_pre']} train rows, "
          f"{dedup_info['rows_removed_post']} test rows")

    df_trainval = df_pre_split.reset_index(drop=True)
    df_test     = df_post_split.reset_index(drop=True)

    feat_cols = [c for c in df_trainval.columns if c not in ("source_file", "gnra")]

    # ── v9 FIX: tune Optuna on THIS split's de-duplicated pool, not once ───────
    # globally on the raw pool. Previously a single Optuna study ran once,
    # before this loop, on df_pre_cutoff BEFORE de-duplication — so the CV
    # folds it tuned against could still contain near-identical duplicate
    # motifs (the very thing de-duplication exists to remove), and whatever
    # HP came out of that one study was reused for all 6 splits regardless of
    # how different each split's deduplicated training distribution was.
    # Now: de-dup already happened above (`df_trainval`), so this study tunes
    # on exactly the distribution the final model for THIS split will train
    # on — no duplicate-driven distribution mismatch, and no cross-split HP
    # reuse.
    split_label = f"{mode} / {method}"
    print(f"  Running Optuna study for this split "
          f"({OPTUNA_N_TRIALS} trials, {HPO_N_SPLITS}-fold CV, on de-duplicated data)…")
    HP = run_optuna_for_split(
        df_trainval, feat_cols, NODE_DIM, EDGE_DIM,
        split_label=split_label,
    )
    HP_by_split[(mode, method)] = HP
    print(f"  HP for this split: {HP}")

    # ── v8 FIX: carve a validation set out of TRAIN, never out of TEST ─────────
    # Previously, `test_loader` (the genuinely held-out post-cutoff data) was
    # evaluated every epoch to pick the "best" checkpoint (early stopping on
    # test MCC), and then the SAME test set was reused to report the final
    # score. That guarantees inflated numbers regardless of how clean the
    # train/test split is — you're reporting the score of the one epoch that
    # best fit the test set, not a genuine generalization estimate. This is
    # almost certainly why de-duplication (v6/v7) had zero effect on the 1.00
    # scores: the leak was in model selection, not in row overlap.
    #
    # Fix: split df_trainval into an internal train/val set (stratified by
    # label). `val_loader` is used for early stopping / checkpoint selection.
    # `test_loader` is built from df_test and is touched exactly once, after
    # training is completely finished, purely for the final reported metrics.
    df_fit, df_val = train_test_split(
        df_trainval,
        test_size=0.15,
        stratify=df_trainval["gnra"],
        random_state=SEED,
    )
    df_fit = df_fit.reset_index(drop=True)
    df_val = df_val.reset_index(drop=True)

    print(f"  Fit  : {len(df_fit)} samples  "
          f"({df_fit['gnra'].sum()} pos / {(~df_fit['gnra']).sum()} neg)")
    print(f"  Val  : {len(df_val)}  samples  "
          f"({df_val['gnra'].sum()} pos / {(~df_val['gnra']).sum()} neg)")
    print(f"  Test : {len(df_test)}  samples  "
          f"({df_test['gnra'].sum()} pos / {(~df_test['gnra']).sum()} neg)  "
          f"[not touched until final evaluation]")

    # ── Scale features — fit ONLY on the fit set ───────────────────────────────
    scaler  = StandardScaler()
    df_fit  = df_fit.copy()
    df_val  = df_val.copy()
    df_test = df_test.copy()
    df_fit[feat_cols]  = scaler.fit_transform(df_fit[feat_cols])
    df_val[feat_cols]  = scaler.transform(df_val[feat_cols])
    df_test[feat_cols] = scaler.transform(df_test[feat_cols])

    # ── Convert rows to graphs ─────────────────────────────────────────────────
    fit_dataset  = [row_to_graph(row, feat_cols) for _, row in df_fit.iterrows()]
    val_dataset  = [row_to_graph(row, feat_cols) for _, row in df_val.iterrows()]
    test_dataset = [row_to_graph(row, feat_cols) for _, row in df_test.iterrows()]

    fit_loader  = DataLoader(fit_dataset,  batch_size=HP["batch_size"], shuffle=True)
    val_loader  = DataLoader(val_dataset,  batch_size=HP["batch_size"])
    test_loader = DataLoader(test_dataset, batch_size=HP["batch_size"])

    # ── Class-weighted loss (computed from the fit set only) ───────────────────
    n_pos = int(df_fit["gnra"].sum())
    n_neg = int((~df_fit["gnra"]).sum())
    pos_weight = n_neg / n_pos if n_pos > 0 else 1.0
    print(f"  pos_weight: {pos_weight:.2f}x  (neg {n_neg} / pos {n_pos})")

    class_weights = torch.tensor([1.0, pos_weight], dtype=torch.float)
    criterion = torch.nn.CrossEntropyLoss(weight=class_weights)

    # ── Model + optimiser ──────────────────────────────────────────────────────
    set_all_seeds(SEED)
    model     = GATModel(HP, NODE_DIM, EDGE_DIM)
    optimizer = torch.optim.Adam(
        model.parameters(), lr=HP["lr"], weight_decay=HP["weight_decay"]
    )

    best_mcc   = -2.0
    best_state = None
    no_improve = 0

    # History buffers — filled each epoch for the training chart.
    # These now track VALIDATION metrics, not test metrics.
    history = {m: [] for m in ("accuracy", "precision", "recall", "f1", "mcc")}

    # ── Training loop with early stopping on VALIDATION MCC ────────────────────
    for epoch in range(1, HP["max_epochs"] + 1):
        train_epoch(model, fit_loader, optimizer, criterion)
        _, val_preds, val_labels = evaluate(model, val_loader)

        m = compute_metrics(val_labels, val_preds)
        for key, val in m.items():
            history[key].append(val)

        val_mcc = m["mcc"]
        if val_mcc > best_mcc:
            best_mcc   = val_mcc
            best_state = copy.deepcopy(model.state_dict())
            no_improve = 0
        else:
            no_improve += 1

        if epoch % 20 == 0:
            print(f"  Epoch {epoch:03d} | val MCC {val_mcc:.4f} (best {best_mcc:.4f})")

        if no_improve >= HP["patience"]:
            print(f"  Early stopping at epoch {epoch} (best val MCC {best_mcc:.4f})")
            break

    # ── Save training chart (validation curves) ────────────────────────────────
    chart_path  = save_training_chart(history, split_label)
    print(f"  Chart saved → {chart_path}")

    # ── Final evaluation — TEST SET TOUCHED HERE, EXACTLY ONCE ─────────────────
    model.load_state_dict(best_state)
    _, final_preds, final_labels = evaluate(model, test_loader)
    final_metrics = compute_metrics(final_labels, final_preds)

    print(f"\n  Classifier: GAT  (split: {split_label})  — held-out TEST metrics")
    print(classification_report(final_labels, final_preds))

    summary_rows.append({
        "split": split_label,
        **final_metrics,
    })

    # ── Save model ─────────────────────────────────────────────────────────────
    model_name = "gat"
    payload = {
        "classifier_name": "GAT",
        "model_state_dict": best_state,
        "hp":              HP,   # this split's own Optuna-tuned HP (v9)
        "scaler":          scaler,
        "feature_columns": feat_cols,
        "node_dim":        NODE_DIM,
        "edge_dim":        EDGE_DIM,
        "num_nodes":       NUM_NODES,
        "window_size":     NUM_NODES,
        "nt_number":       NTnumber,   # explicit name matching predict.py's --nt-number CLI arg
        "positive_label":  True,
        "split_mode":      mode,
        "split_method":    method,
    }
    out_path = f"{mode}-{method}-{model_name}_v10.pkl"
    with open(out_path, "wb") as f:
        dump(payload, f)
    print(f"  Saved → {out_path}")

# ── Print cross-split summary ──────────────────────────────────────────────────
print("\n" + "="*60)
print("  CROSS-SPLIT SUMMARY (held-out test set, never used for model selection)")
print("="*60)
summary_df = pd.DataFrame(summary_rows).set_index("split")
summary_df = summary_df[["accuracy", "precision", "recall", "f1", "mcc"]]
print(summary_df.to_string(float_format=lambda x: f"{x:.4f}"))
print("\nMean across splits:")
print(summary_df.mean().to_string(float_format=lambda x: f"{x:.4f}"))
```

## Label-Shuffle Sanity Check

**Why this cell exists:** every previous version of this notebook (v6, v7) reported
near-perfect scores that turned out to be caused by a leakage bug, not real signal.
v8 fixes the known leaks (per-split de-duplication + test set touched exactly once),
but the held-out test MCC is *still* 0.94-1.00 across splits, including two splits at
a flat 1.0000. That pattern is exactly what the earlier bugs also looked like, so it
deserves one more check before trusting the numbers.

**The check:** re-run the *identical* pipeline (same de-dup, same fit/val/test split
logic, same architecture, same early-stopping rule, and — since v9 — that split's own
Optuna-tuned HP from `HP_by_split`, not a single shared HP dict) but with the `gnra`
labels randomly permuted within each pool before training. Permuting breaks any real
relationship between the geometric features and the label while leaving everything
else — cluster structure, class balance per pool, training procedure — untouched.

- If shuffled-label test MCC collapses to ~0 → the real-label MCC is measuring a
  genuine, learnable geometric signature (plausible for GNRA tetraloops, which have
  a very stereotyped 3D backbone conformation).
- If shuffled-label test MCC stays well above 0 → the model can hit high performance
  regardless of what the label actually says, which means some feature in
  `geometric_features.csv` is likely acting as a proxy for the label itself
  (e.g. the same geometric threshold rule used to originally annotate `gnra`).
  In that case the fix is upstream — auditing/removing the tautological feature —
  not another split-logic fix.

By default this runs on 2 of the 6 splits to keep runtime reasonable; set
`SHUFFLE_CHECK_SPLITS = list(clustering.keys())` to run all six.


```python
def run_split_experiment(
    mode, method, clusters, df_pre_cutoff, df_post_cutoff,
    hp, node_dim, edge_dim, shuffle_labels=False, seed=SEED, verbose=True,
):
    """
    Runs the exact same per-split pipeline as the Main Training Loop (Cell 24):
    de-dup -> stratified fit/val split out of the train pool -> scale (fit on
    fit-set only) -> graph encode -> train with early stopping on VAL MCC ->
    single evaluation on the held-out test set.

    If shuffle_labels=True, the 'gnra' column is randomly permuted *within
    each pool* (pre-cutoff and post-cutoff separately) before anything else
    happens. This preserves each pool's class balance and leaves cluster
    membership, dates, and features completely untouched -- the only thing
    that changes is which label is attached to which row. Any performance
    that survives this permutation cannot be coming from real feature/label
    structure.
    """
    rng = np.random.default_rng(seed)

    df_pre_s  = df_pre_cutoff.copy()
    df_post_s = df_post_cutoff.copy()

    if shuffle_labels:
        df_pre_s["gnra"]  = rng.permutation(df_pre_s["gnra"].values)
        df_post_s["gnra"] = rng.permutation(df_post_s["gnra"].values)

    df_pre_split, df_post_split, dedup_info = dedup_clusters_for_split(
        df_pre_s, df_post_s, clusters, n_keep=1, seed=seed
    )

    df_trainval = df_pre_split.reset_index(drop=True)
    df_test     = df_post_split.reset_index(drop=True)

    df_fit, df_val = train_test_split(
        df_trainval, test_size=0.15, stratify=df_trainval["gnra"], random_state=seed
    )
    df_fit = df_fit.reset_index(drop=True)
    df_val = df_val.reset_index(drop=True)

    feat_cols = [c for c in df_fit.columns if c not in ("source_file", "gnra")]

    scaler  = StandardScaler()
    df_fit  = df_fit.copy()
    df_val  = df_val.copy()
    df_test = df_test.copy()
    df_fit[feat_cols]  = scaler.fit_transform(df_fit[feat_cols])
    df_val[feat_cols]  = scaler.transform(df_val[feat_cols])
    df_test[feat_cols] = scaler.transform(df_test[feat_cols])

    fit_dataset  = [row_to_graph(row, feat_cols) for _, row in df_fit.iterrows()]
    val_dataset  = [row_to_graph(row, feat_cols) for _, row in df_val.iterrows()]
    test_dataset = [row_to_graph(row, feat_cols) for _, row in df_test.iterrows()]

    fit_loader  = DataLoader(fit_dataset,  batch_size=hp["batch_size"], shuffle=True)
    val_loader  = DataLoader(val_dataset,  batch_size=hp["batch_size"])
    test_loader = DataLoader(test_dataset, batch_size=hp["batch_size"])

    n_pos = int(df_fit["gnra"].sum())
    n_neg = int((~df_fit["gnra"]).sum())
    pos_weight = n_neg / n_pos if n_pos > 0 else 1.0
    class_weights = torch.tensor([1.0, pos_weight], dtype=torch.float)
    criterion = torch.nn.CrossEntropyLoss(weight=class_weights)

    set_all_seeds(seed)
    model = GATModel(hp, node_dim, edge_dim)
    optimizer = torch.optim.Adam(model.parameters(), lr=hp["lr"], weight_decay=hp["weight_decay"])

    best_mcc, best_state, no_improve = -2.0, None, 0

    for epoch in range(1, hp["max_epochs"] + 1):
        train_epoch(model, fit_loader, optimizer, criterion)
        _, val_preds, val_labels = evaluate(model, val_loader)
        val_mcc = matthews_corrcoef(val_labels, val_preds)

        if val_mcc > best_mcc:
            best_mcc, best_state, no_improve = val_mcc, copy.deepcopy(model.state_dict()), 0
        else:
            no_improve += 1
        if no_improve >= hp["patience"]:
            break

    model.load_state_dict(best_state)
    _, final_preds, final_labels = evaluate(model, test_loader)
    final_metrics = compute_metrics(final_labels, final_preds)

    if verbose:
        tag = "SHUFFLED" if shuffle_labels else "real    "
        print(f"  [{tag}] {mode}/{method}  "
              f"fit={len(df_fit)} val={len(df_val)} test={len(df_test)}  "
              f"best_val_mcc={best_mcc:.4f}  test_mcc={final_metrics['mcc']:.4f}  "
              f"test_f1={final_metrics['f1']:.4f}")

    return final_metrics


# ── Run the shuffle check ────────────────────────────────────────────────────
# Full 6-split loop takes a while, so by default this checks a subset;
# set SHUFFLE_CHECK_SPLITS = list(clustering.keys()) to run all six.
SHUFFLE_CHECK_SPLITS = [("approximate", "hierarchical"), ("exact", "affinity-propagation")]

print("Running label-shuffle sanity check (identical pipeline, permuted 'gnra' labels)...\n")

shuffle_results = []
for (mode, method) in SHUFFLE_CHECK_SPLITS:
    clusters = clustering[(mode, method)]

    # v9: use THIS split's own Optuna-tuned HP (from the Main Training Loop),
    # not a single shared HP dict — keeps this check consistent with how the
    # split was actually trained. Falls back to HP_DEFAULT + fixed schedule
    # settings if the main loop wasn't run for this split for some reason.
    split_hp = HP_by_split.get((mode, method))
    if split_hp is None:
        print(f"  [warning] no tuned HP found for {mode}/{method}; "
              f"falling back to HP_DEFAULT.")
        split_hp = {
            **HP_DEFAULT,
            "batch_size": HP_BATCH,
            "max_epochs": FINAL_MAX_EPOCHS,
            "patience":   FINAL_PATIENCE,
        }

    real_metrics = run_split_experiment(
        mode, method, clusters, df_pre_cutoff, df_post_cutoff,
        split_hp, NODE_DIM, EDGE_DIM, shuffle_labels=False, seed=SEED,
    )
    shuf_metrics = run_split_experiment(
        mode, method, clusters, df_pre_cutoff, df_post_cutoff,
        split_hp, NODE_DIM, EDGE_DIM, shuffle_labels=True, seed=SEED,
    )

    shuffle_results.append({
        "split": f"{mode}/{method}",
        "real_mcc": real_metrics["mcc"],
        "shuffled_mcc": shuf_metrics["mcc"],
    })
    print()

shuffle_df = pd.DataFrame(shuffle_results).set_index("split")
print(shuffle_df.to_string(float_format=lambda x: f"{x:.4f}"))
print("\nInterpretation:")
print(" - shuffled_mcc near 0        -> real_mcc reflects genuine, learnable signal.")
print(" - shuffled_mcc still >> 0    -> some feature likely encodes the label directly;")
print("                                 audit feature provenance / how 'gnra' was annotated.")
```
